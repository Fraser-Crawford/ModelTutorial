# NumPy
Numpy is a python library that makes operating on large arrays of numbers much easier. It also provides a lot of useful functions for math operations.

The basic object is the numpy array. Unlike a list, these have a set size and all the data within them must be the same type. Unlike basic python, numpy has more datatypes. Some operations work very differently...

```python
import numpy

array_1 = numpy.array([1,2,3])

print(array_1*2) #Most operations are performed element wise

print(array_1*array_1) #You can also operate an array with an array element wise as well
```

Arrays can be any dimensional, they are defined by a shape:

```python
import numpy as np #shorthand for numpy commonly used
ones_array = np.ones((3,6)) #3 rows, 6 columns. Its like a 3 element list where each element is a 6 element list.
print(ones_array)
```

You may need to do operations on a particular axis of the array. For example, the array above might represent the mass of each component in a 6 component mixture in 3 set ups. If I wanted the total mass in each set up, I need to add up all the columns into a single column. This would turn the array from (3,6) to (3,1). Numpy by default will get rid of the single length dimension (unless keep_dimension is True) so the result is an array of shape (3,). This is an operation of axis 1.

```python
summed_array = np.sum(ones_array)
summed_rows = np.sum(ones_array, axis=0)
summed_columns = np.sum(ones_array, axis=1)

print(summed_array, summed_rows, summed_columns)

summed_array = np.sum(ones_array, keepdims=True)
summed_rows = np.sum(ones_array, axis=0, keepdims=True)
summed_columns = np.sum(ones_array, axis=1, keepdims=True)

print(summed_array)
print(summed_rows)
print(summed_columns)
```
