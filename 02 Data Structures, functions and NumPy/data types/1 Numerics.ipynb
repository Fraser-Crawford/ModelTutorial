# Data Structures

All data is stored as binary data on a computer. Data types allow the interpreter/compiler to read this data into a meaningful value for further operations. This section details the use cases of these structures and some useful tips for making the best use of them.

# Numeric Types

In base Python you have only 2/3 options:
- Integers, whole numbered values with no bound. Both positive and negative.
- Floats, a method for representing decimal numbers. They represent numbers as a mantissa and a exponent like this: $\textrm{Mantissa} \times 2^{\textrm{Exponent}}$. This means that the gaps between numbers grows as the float value grows! For large floats, you can get weird behaviour such as N+1=N as the gap from N to the next float is much larger than 1.
- Booleans, a weird case, represent values that a are True or False. 0 is False, all other numbers are True. However if you use True in an equation, it's equal to 1.

```python
integer_1 = 9
integer_2 = 10

print(integer_1/integer_2) #Division in modern python always produces a float value

#Float and integers can be converted using the int() or float() functions:

print(int(integer_1/integer_2))
```

```python
print(integer_1//integer_2) #Integer division is performed using a // symbol. This is the same as converting the resulting float to an integer if the result was positive.
print(int(integer_1/integer_2))
```

```python
large_float = 1e63 #Can use scientific notation to define floats. Even if the result could be an integer, this will always result in a float.
print(large_float)
print(large_float-(large_float+1)) #The values end up identical after the addition!
```

# Operations

Useful operations:
- +, -, *, / for normal arithmatic
- x**y is x to the power of y
- % gets the remainder from a division (5 % 2 = 3)
- == tells you if two values are equal. It may be better for float arithmatic to use math.isclose(x,y) when using really small or large numbers.
- You can add an operator before an assignment equal sign to apply the operation in place


```python
i = 1
i += 1
print(i)
i *= 2
print(i)
i /= 2
print(i)
i **= 2
print(i)
```
