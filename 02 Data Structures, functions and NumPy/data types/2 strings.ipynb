# Strings

Strings are a datatype for text. Python treats them like a list of characters. They can be defined by "" or ''. There are some special characters such as \n for newline.

```python
normal_string = "Hello there"
print(normal_string)
```

When you print something, it gets converted to a string. This can also be done via the str() function.

```python
value = 1.0
string_value = str(value)
print(value, string_value)
```

A small amount of math operations are allowed on strings: + and *.

```python
string_value = "123"
add_string = string_value + "1" # Concatenates 1 onto the end
print(add_string)
multiply_string = string_value*3 # Duplicates
print(multiply_string)
```

It may be useful to shorten values that have been converted to strings such as decimals

```python
third = 1/3
print(third)

print(format(third, '.2f')) # 2 decimal places
print(f"{third:.2f}") # Equivalent form
```

Strings can be sliced and accessed just like lists

```python
long_string = "The quick fox jumped over the lazy dog"

print(long_string[:7])
print(long_string[::-1])
```

You can also split a string into a list of strings or join a list of strings into a single string:

```python
words = long_string.split(" ") #split at the spaces
print(words)
reverse_words = words[::-1]
backwards_words_string = " ".join(reverse_words) #join with spaces
print(backwards_words_string)
```
