# Lists
Lists are an ordered collection of data types. List elements are accessed via an index which starts at 0 for the 0th (1st) element. Negative indices can be used to traverse the list backwards.

They are defined using square brackets.

```python
list_1 = [0,1,2,3,4,5,6]

print(list_1[4])
print(list_1[-1]) #-1 is the last element, -2 the second to last etc.
```

Lists can be sliced to make sub lists using a [ start : stop : step ] syntax. A blank start is 0, a blank stop is the end of the list.

```python
even_values = list_1[::2]
print(even_values)
reverse_list = list_1[::-1]
print(reverse_list)
```

You can edit a particular element in place. Note that this modifies the element in all lists that point to the same data!

```python
duplicate_list = list_1
duplicate_list[2] = 500
print(duplicate_list)
print(list_1)
```

You can also modify slices of lists in place

```python
duplicate_list[2:3] = ["Hello", "there"]
print(duplicate_list)
```

To make a copy of a list without it sharing the same data, use a full slice, [:]

```python
duplicate_list = list_1[:]
duplicate_list[2] = 500
print(duplicate_list)
print(list_1)
```

Lists can contain other lists! These are referred to as 2D lists.

```python
list_2D = [[0,1],[2,3]]
print(list_2D[0]) #The first element is the first list in the 2D list
```

But when you copy a 2D list, the contained lists of the copy still point to the same original data! Use a deepcopy to avoid odd behaviour.

```python
duplicate_list = list_2D[:]
list_2D[1][1] += 100 #Open the first list, edit the 1st element of that list

print(duplicate_list)
print(list_2D)

from copy import deepcopy #importing deepcopy
list_2D = [[0,1],[2,3]]
duplicate_list = deepcopy(list_2D)
list_2D[1][1] += 100

print(duplicate_list)
print(list_2D)
```

Lists can be made using iterators like range:

```python
list_1 = list(range(5)) #range is a useful function for making number sequences
print(list_1)
```

You can also use comprehension syntax:

```python
list_1 = [2*x for x in range(5)]
print(list_1)
```

Elements can be removed by index or value in a couple ways alongside insertion:

```python
list_1 = [1,2,3,4,"banana",5,6]
print(list_1.pop(0)) #Pop removes the value at the index and returns it. If no index is given, the final index is used.
print(list_1.remove("banana")) #Remove the first item from the list with this value. Doesn't return anything
print(list_1.insert(0,"banana")) #Insert the value at the index, shift everything to the right. Doesn't return anything
print(list_1)
```

Lists can be sorted in two ways, in place and as a new list

```python
list_1 = [5,23423,87,234,346,6855]
print(sorted(list_1))
print(list_1)
print(list_1.sort()) #Doesn't return anything
print(list_1)
```

Lists can be added to via appending or extending:

```python
list_1 = [1,2,3]
list_2 = [4,5,6]

print(list_1 + list_2)

list_1 = [1,2,3]
list_2 = [4,5,6]

print(list_1.extend(list_2)) #Same as above
print(list_1)

list_1 = [1,2,3]
list_2 = [4,5,6]

print(list_1.append("banana"))
print(list_1)
```
