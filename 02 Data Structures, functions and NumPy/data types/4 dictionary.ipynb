# Dictionaries

Unlike lists that are accessed via an ordered integer index, dictionaries associate unique keys with values. This makes them very descriptive. They are created using curly braces {} as a key : value pair

```python
dictionary_1 = {"key":"value"}
print(dictionary_1["key"])
```

You may try to access a key that doesn't exist yet. To avoid a crash, use the get(key) function on the dictionary.

```python
print(dictionary_1.get("key"))
print(dictionary_1.get("banana"))
```
