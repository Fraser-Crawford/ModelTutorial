# Functions
A function is a reusable piece of code that takes in some values and does something. It can also return some values as well.

```python
def add_3_numbers(input_1, input_2, input_3):
    return input_1 + input_2 + input_3

print(add_3_numbers(1, 2, 3))
```

The function can use default values for common inputs. The default values must come after all the necessary ones.

```python
def add_3_numbers(input_1, input_2, input_3=1):
    return input_1 + input_2 + input_3

print(add_3_numbers(1, 2))
```

Functions can be stored like a variable for very complex behaviour!

```python
def make_adding_function(number_to_add):
    def add(x):
        return x + number_to_add
    return add

adding_function = make_adding_function(3)
print(adding_function(4))
```

You can add hints to the function to help inform the user what kind of data it expects. These hints are not enforced, you're allowed to break them.

```python
def add_3_numbers(input_1:float, input_2:float, input_3:float=1)->float:
    return input_1 + input_2 + input_3

print(add_3_numbers(1, 2))
```
